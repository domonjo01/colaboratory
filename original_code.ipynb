```python
import numpy as np

# データ：身長
matrix = np.array([165, 160, 166, 164, 168, 164,168, 169,169, 166],)

# データをソート：昇順
sorted_matrix = np.sort(matrix)

# 偏差
deviation = sorted_matrix - np.mean(matrix)

# 偏差の二乗
deviation_squared = deviation ** 2

# 平均
average = np.sum(sorted_matrix) / len(sorted_matrix)

# 偏差の2乗の和
sum_deviation_squared = np.sum(deviation_squared)

# 分散：偏差の和 / 個数（偏差の和の平均）
variance= np.sum(sum_deviation_squared) / len(sorted_matrix)

# 標準偏差
standard_deviation = np.sqrt(variance)

# 出力
print(f"身長： {sorted_matrix}")
print(f"偏差： {deviation}")
print(f"平均： {average}")
print(f"偏差の2乗： {deviation_squared}")
print(f"偏差の2乗の和： {sum_deviation_squared}")
print(f"標準偏差： {standard_deviation}")

# 関数出力
print("平均：" + str(np.mean(sorted_matrix)))
print("分散の和：" + str(np.var(sorted_matrix)))
print("標準偏差：" + str(np.std(sorted_matrix)))
```
